## 1. Виртуальные окружения (venv)

### Что такое виртуальное окружение?

Виртуальное окружение — это изолированная среда для Python-проекта, которая позволяет:
- Устанавливать библиотеки только для конкретного проекта
- Избегать конфликтов версий между разными проектами
- Легко делиться списком зависимостей с другими разработчиками

### Как создать виртуальное окружение?

**Windows:**
```bash
python -m venv my_venv
my_venv\Scripts\activate
```

**macOS/Linux:**
```bash
python3 -m venv my_venv
source my_venv/bin/activate
```

После активации вы увидите `(my_venv)` в начале строки терминала.

## 1. Введение в NumPy

### Что такое NumPy?

**NumPy** (Numerical Python) — это библиотека для работы с многомерными массивами и математических вычислений.

### Зачем нужен NumPy?

- **Скорость**: операции с NumPy в десятки раз быстрее, чем с обычными списками Python
- **Удобство**: простой синтаксис для сложных математических операций
- **Функциональность**: огромное количество встроенных функций

### NumPy vs Python списки:

```python
# Python список
python_list = [1, 2, 3, 4, 5]
doubled = [x * 2 for x in python_list]  # Нужен цикл

# NumPy array
numpy_array = np.array([1, 2, 3, 4, 5])
doubled = numpy_array * 2  # Операция сразу для всех элементов
```

## 2. Индексация: loc vs iloc

### loc - доступ по МЕТКАМ (labels)
- Использует названия строк и столбцов
- Синтаксис: `df.loc[rows, columns]`
- Включает конечное значение в срез

### iloc - доступ по ПОЗИЦИЯМ (integer positions)
- Использует числовые индексы (0, 1, 2, ...)
- Синтаксис: `df.iloc[rows, columns]`
- НЕ включает конечное значение (как обычный Python срез)

### Когда что использовать?
- **loc**: когда знаете название/метку ("дайте строку с датой 2024-01-15")
- **iloc**: когда знаете позицию ("дайте 5-ую строку")

## 4. Сложные условия

### Логические операторы:
- `&` - И (AND) - оба условия должны быть True
- `|` - ИЛИ (OR) - хотя бы одно условие True
- `~` - НЕ (NOT) - инверсия условия

### ВАЖНО:
Каждое условие должно быть в скобках!
```python
# Правильно:
df[(df['Temperature'] > 20) & (df['Humidity'] < 70)]

# Неправильно:
df[df['Temperature'] > 20 & df['Humidity'] < 70]
```

### Пример 7: Условие И (AND)

````python
# Найдите дни с температурой выше 20°C И влажностью ниже 70%
df[~((df['temp'] > 0) | (df['hum'] > 70))]
````

### Пример 8: Условие ИЛИ (OR)

````python
# Найдите дни с очень низкой (< 15°C) ИЛИ очень высокой (> 30°C) температурой
df[(df['temp'] < -7) | (df['temp'] > 2)]
````

### Пример 9: Условие НЕ (NOT)

````python
# Найдите дни без осадков (осадки == 0)
# Затем найдите дни С осадками, используя NOT
df[~(df.temp == 0)]
````

### Пример 10: Проверка на вхождение в список (.isin())

````python
# Найдите дни с определенным описанием погоды
# Например: только 'Ясно' или 'Облачно'
df[df.temp.isin([-10,4,0])]
````
